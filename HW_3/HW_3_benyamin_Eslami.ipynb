**1 [10pts].** Consider a variant of linear regression in which each data point $(\mathbf{x}_i, y_i)$ is associated with a weighting factor $r_i > 0$. The sum-of-square error for a linear function $\theta(\mathbf{x}) = \mathbf{w}^\top\mathbf{x}$ parameterized by $\mathbf{w}$ in this case will be:

$\mathbf{E}(\mathbf{w}) = \sum_{i=1}^n r_i\Big(\mathbf{w}^\top\mathbf{x}_i - y_i\Big)^2$ where $n$ is the number of data points.

(a) Derive the gradient descent rule for this, i.e., derive $\mathrm{d}\mathbf{E}/\mathrm{d}\mathbf{w}$ (5pts).  

**Answer**

Gradient Descent formula:

$$
\lambda^{t+1} \leftarrow \lambda^{t} - \alpha \frac{\partial}{\partial \mathbf w} L(\lambda; X,y)
$$

We take the derivative of $E(\mathbf w)$ with respect to $\mathbf w$:

$
\frac{\partial}{\partial \mathbf w} E(\mathbf w) = \sum_{i=1}^n r_i\Big(\mathbf{w}^\top\mathbf{x}_i - y_i\Big)^2 = \sum_{i=1}^n 2 \cdot r_i\Big(\mathbf{w}^\top\mathbf{x}_i - y_i\Big).\mathbf{x}_i
$

Substituting the derivative to the above GD formula:


$$
\lambda^{t+1} \leftarrow \lambda^{t} - 2 \cdot \alpha \sum_{i=1}^n r_i\Big(\mathbf{w}^\top\mathbf{x}_i - y_i\Big).\mathbf{x}_i
$$

(b) Derive the closed-form solution for the optimal $\mathbf{w}$ that minimizes the above error. Show your work (5pts).

**Answer**

SSE is given. The goal is to find $\mathbf w^* = \mathrm{argmin} E(\mathbf w)$

$$
E(\mathbf w)=\sum_{i=1}^{n}r_i(\mathbf w^\top\mathbf x_i-y_i)^2
$$

For n observations, expanding the summation gives:

$$
E(\mathbf w)= \\
  r_1(\mathbf w^\top\mathbf x_1-y_1)^2\\
+ r_2(\mathbf w^\top\mathbf x_2-y_2)^2\\
+ r_3(\mathbf w^\top\mathbf x_3-y_3)^2 \\
...
$$

$$
\mathbf X=
\begin{bmatrix}
\mathbf x_1\\
\mathbf x_2\\
\vdots\\
\mathbf x_n
\end{bmatrix}
$$

$$
\mathbf y=
\begin{bmatrix}
y_1\\y_2\\\vdots\\y_n
\end{bmatrix}
$$

$$
\mathbf R=
\begin{bmatrix}
r_1&0&\cdots&0\\
0&r_2&\cdots&0\\
\vdots&\vdots&\ddots&\vdots\\
0&0&\cdots&r_n
\end{bmatrix}
$$

Our goal is to express these n terms using matrix operations. Suppose our model has l features. Then each observation has l features
$$
\mathbf w=
\begin{bmatrix}
\mathbf w_1\\
\mathbf w_2\\
\vdots\\
\mathbf w_l
\end{bmatrix}
$$

$$
\mathbf X_i=
\begin{bmatrix}
\mathbf x_{i1}\\
\mathbf x_{i2}\\
\vdots \\
\mathbf x_{in}
\end{bmatrix}
$$

We construct the input matrix $X$ by placing each observatin in a separate tow:
$$
\mathbf X=
\begin{bmatrix}
x_{11}&x_{12}&...&x_{1l}\\
x_{21}&x_{22}&...&x_{2l}\\
\vdots \\
x_{n1}&x_{n2}&...&x_{nl}
\end{bmatrix}
$$

Now multiply $\mathbf X$ by $\mathbf w$

$$
\mathbf X\mathbf w=
\begin{bmatrix}
x_{11}w_1+x_{12}w_2+...+x_{1l}w_l\\
x_{21}w_1+x_{22}w_2+...+x_{2l}w_l\\
\vdots \\ 
x_{n1}w_1+x_{n2}w_2+...+x_{nl}w_l\\
\end{bmatrix}
=
\begin{bmatrix}
\mathbf w^\top\mathbf x_1\\
\mathbf w^\top\mathbf x_2\\
\vdots \\
\mathbf w^\top\mathbf x_n
\end{bmatrix}
$$

If we subtract this from the prediction vector:

$$
\mathbf X\mathbf w-\mathbf y=
\begin{bmatrix}
\mathbf w^\top\mathbf x_1-y_1\\
\mathbf w^\top\mathbf x_2-y_2\\
\vdots \\
\mathbf w^\top\mathbf x_n-y_n
\end{bmatrix}
$$

We know that each observation has its own weight $r_i$. We already parametrized $R$ matrix. So, we mulitply it

$$
\mathbf R (\mathbf X\mathbf w-\mathbf y)=
\begin{bmatrix}
r_1&0&\cdots&0\\
0&r_2&\cdots&0\\
\vdots&\vdots&\ddots&\vdots\\
0&0&\cdots&r_n
\end{bmatrix}
\begin{bmatrix}
\mathbf w^\top\mathbf x_1-y_1\\
\mathbf w^\top\mathbf x_2-y_2\\
\vdots \\
\mathbf w^\top\mathbf x_n-y_n
\end{bmatrix}
=
\begin{bmatrix}
r_1(\mathbf w^\top\mathbf x_1-y_1)\\
r_2(\mathbf w^\top\mathbf x_2-y_2)\\
\vdots \\
r_n(\mathbf w^\top\mathbf x_n-y_n)
\end{bmatrix}
$$
We want the sum of the weighted squared errors. So we must multiply $\mathbf R (\mathbf X\mathbf w-\mathbf y)$ by $(\mathbf X\mathbf w-\mathbf y)^T$.
This will give us exactly the original weighted sum of squared errors. Therefore,

$$
E(\mathbf w)= (\mathbf X\mathbf w-\mathbf y)^\top \mathbf R (\mathbf X\mathbf w-\mathbf y)
$$

Now if we expand the above formula and take the derivative with respect to $\mathbf w$

$$
E(\mathbf w)= \mathbf w^\top \mathbf X^\top \mathbf R \mathbf X \mathbf w - 2 \mathbf W^\top \mathbf X^\top \mathbf R \mathbf y + \mathbf y^\top \mathbf R \mathbf y \\

E(\mathbf w)= ||\mathbf w||^2 \mathbf X^\top \mathbf R \mathbf X \mathbf w - 2 \mathbf W^\top \mathbf X^\top \mathbf R \mathbf y + \mathbf y^\top \mathbf R \mathbf y \\

\frac{\partial E(\mathbf w)}{\partial\mathbf w} = 2\mathbf X^\top\mathbf R\mathbf X\mathbf w - 2\mathbf X^\top\mathbf R\mathbf y = 0 \\
 
2\mathbf X^\top\mathbf R\mathbf X\mathbf w = 2\mathbf X^\top\mathbf R\mathbf y
$$

multiply both sides by $X^\top\mathbf R\mathbf X$

$$
\mathbf w^* = (\mathbf X^\top\mathbf R\mathbf X)^{-1} \mathbf X^\top\mathbf R\mathbf y
$$

**2 [10pts].** Consider a maximum likelihood estimation (MLE) problem with the following probabilistic model:

$y_i = \mathbf{w}^\top \mathbf{x}_i + \epsilon_i$ where $\epsilon_i \sim \mathbb{N}\left(0, \frac{1}{r_i}\right)$ where $1 \leq i \leq n$.

Show that this MLE model is equivalent to the above linear regression model with weighting factors. You can do that by following the below steps.

(a) Find an analytically computable form for the likelihood of data given $\mathbf{w}$, $\mathbf{L}(\mathbf{w}) \triangleq P(y_1, y_2, \ldots, y_n \mid \mathbf{x}_1, \mathbf{x}_2, \ldots, \mathbf{x}_n)$. Show your work (5pts).

**Answer**

Since the noise follows a Gaussian distribution with mean zero and variance $\frac{1}{r_i}$, adding the deterministic quantity $\mathbf w^\top\mathbf x_i$ shifts its mean without changing its variance. Therefore, 

$$
y_i \mid \mathbf x_i,\mathbf w \sim \mathcal N\left(\mathbf w^\top\mathbf x_i,\frac{1}{r_i}\right)
$$

The probability density function (PDF) of gaussian distribution is

$$
p(y)= \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left(-\frac{(y-\mu)^2}{2\sigma^2}\right)
$$

For our problem, $\mu_i=\mathbf w^\top\mathbf x_i, \quad \sigma_i^2=\frac{1}{r_i}$. We substitute them into the PDF formula:

$$
p(y_i\mid\mathbf x_i,\mathbf w) = \frac{1}{\sqrt{2\pi/r_i}} \exp\left(-\frac{(y_i-\mathbf w^\top\mathbf x_i)^2}{2/r_i} \right)
$$

The above formula is the liklihood function. Now that we want to calculate joint liklihood, we have $L(\mathbf w)= \prod_{i=1}^{n}p(y_i\mid\mathbf x_i,\mathbf w)$. We express this multiplication by using PDF function:

$$
L(\mathbf w)= \prod_{i=1}^{n} \left[ \sqrt{\frac{r_i}{2\pi}}\exp\left( -\frac{r_i}{2}(y_i-\mathbf w^\top\mathbf x_i)^2 \right) \right]
$$

We can simplify the formula by separating the product into two parts:

$$
L(\mathbf w)= \left(\prod_{i=1}^{n}\sqrt{\frac{r_i}{2\pi}}\right) \left(\prod_{i=1}^{n} \exp\left[-\frac{r_i}{2}(y_i-\mathbf w^\top\mathbf x_i)^2\right]\right)
$$

Since we have exponential function in the PDF formula, using its property, we can convert the product into summation:

$$
L(\mathbf w)= \left(\prod_{i=1}^{n}\sqrt{\frac{r_i}{2\pi}}\right) \exp\left(-\frac12\sum_{i=1}^{n}r_i(y_i-\mathbf w^\top\mathbf x_i)^2 \right)
$$

We can simplify the formula by chaning $\prod_{i=1}^{n}\sqrt{\frac{r_i}{2\pi}} = (2\pi)^{-n/2}\prod_{i=1}^{n}\sqrt{r_i}$. Therefore, the final liklihood function will become:

$$
L(\mathbf w)= (2\pi)^{-n/2} \left(\prod_{i=1}^{n}\sqrt{r_i}\right) \exp\left(-\frac12\sum_{i=1}^{n} r_i(\mathbf w^\top\mathbf x_i-y_i)^2 \right)
$$

(b) Find $\mathbf{E}(\mathbf{w}) = -log\ \mathbf{L}(\mathbf{w})$ and the optimal $\mathbf{w}$ (in closed-form) that minimizes $\mathbf{E}(\mathbf{w})$. Show your work. You shoud see that the closed-form solution for $\mathbf{w}$ here is the same as that of **1.** above (5pts).

**Answer**
$$
E(\mathbf w) = -\log\left[\prod_{i=1}^{n} \sqrt{\frac{r_i}{2\pi}} \exp\left(-\frac{r_i}{2}(\mathbf w^\top\mathbf x_i-y_i)^2 \right) \right] 
$$

$$
E(\mathbf w) = -\sum_{i=1}^{n} \left[\frac12\log\left(\frac{r_i}{2\pi}\right) -\frac{r_i}{2}(\mathbf w^\top\mathbf x_i-y_i)^2 \right]
$$

$$
E(\mathbf w) = \frac12\sum_{i=1}^{n} r_i(\mathbf w^\top\mathbf x_i-y_i)^2 + \frac n2\log(2\pi) - \frac12\sum_{i=1}^{n}\log r_i
$$

Since $r_i$ and $n$ are constants independent of $\mathbf w$, the last two terms do not influence the optimal solution. therefore, minimizing the negative log-liklihood is equivalent to minimizing

$$
E(\mathbf w) = \frac12\sum_{i=1}^{n} r_i(\mathbf w^\top\mathbf x_i-y_i)^2
$$

this is exactly one-half of the weighted squared-error function from Question 1. Multiplying an objective by a positive constant does not change its minimizer.

$$
E(\mathbf w) = \frac12(\mathbf X\mathbf w-\mathbf y)^\top \mathbf R(\mathbf X\mathbf w-\mathbf y) = \frac12\left[\mathbf w^\top\mathbf X^\top\mathbf R\mathbf X\mathbf w -2\mathbf w^\top\mathbf X^\top\mathbf R\mathbf y +\mathbf y^\top\mathbf R\mathbf y \right] \\

\nabla_{\mathbf w}E'(\mathbf w) = \mathbf X^\top\mathbf R\mathbf X\mathbf w - \mathbf X^\top\mathbf R\mathbf y \\

\mathbf w^* = (\mathbf X^\top\mathbf R\mathbf X)^{-1} \mathbf X^\top\mathbf R\mathbf y
$$

The last formula is exactly the closed-form solution obtained in Question 1. Weighted Least Squares is equivalent to Maximum Likelihood Estimation when the observations have independent Gaussian noise.

```python
## Libs
import numpy as np 
import pandas as pd
from pathlib import Path
from sklearn.naive_bayes import BernoulliNB
```

```python
data_path = "fortune-cookie-data/"
stop_list_path = data_path + "stoplist.txt"
test_data_path = data_path + "testdata.txt"
test_labels_path = data_path + "testlabels.txt"
train_data_path = data_path + "traindata.txt"
train_labels_path = data_path + "trainlabels.txt"

stop_list = {
    line.strip()
    for line in Path(stop_list_path).read_text(encoding="utf-8").splitlines()
    if line.strip()
}

test_data = [
    line.strip()
    for line in Path(test_data_path).read_text(encoding="utf-8").splitlines()
    if line.strip()
]

train_data = [
    line.strip()
    for line in Path(train_data_path).read_text(encoding="utf-8").splitlines()
    if line.strip()
]

test_labels = np.array([
    int(line.strip())
    for line in Path(test_labels_path).read_text(encoding="utf-8").splitlines()
    if line.strip()
])

train_labels = np.array([
    int(line.strip())
    for line in Path(train_labels_path).read_text(encoding="utf-8").splitlines()
    if line.strip()
])

if len(train_data) != len(train_labels) or len(test_data) != len(test_labels):
    raise ValueError("Each message file must have the same number of lines as its label file")
if not set(train_labels).issubset({0, 1}) or not set(test_labels).issubset({0, 1}):
    raise ValueError("Labels must be binary: 0 or 1")

def tokenize(message: str, stopwords: set[str]) -> set[str]:
    return {word for word in message.split() if word not in stopwords}

def binary_bag_of_words(messages: list[str], vocabulary: list[str], stopwords: set[str],) -> np.ndarray:
    vocabulary_index = {word: index for index, word in enumerate(vocabulary)}
    features = np.zeros((len(messages), len(vocabulary)), dtype=np.float64)
    for row, message in enumerate(messages):
        for word in tokenize(message, stopwords):
            index = vocabulary_index.get(word)
            if index is not None:
                features[row, index] = 1
    return features
    
vocabulary = sorted({word for message in train_data for word in tokenize(message, stop_list)})
train_features = binary_bag_of_words(train_data, vocabulary, stop_list)
test_features = binary_bag_of_words(test_data, vocabulary, stop_list)
```

```python
class BernoulliNaiveBayes:
    def __init__(self, alpha: float = 1.0):
        if alpha <= 0:
            raise ValueError("alpha must be positive")
        self.alpha = alpha

    def fit(self, features: np.ndarray, labels: np.ndarray) -> "BernoulliNaiveBayes":
        self.classes_ = np.array([0, 1], dtype=int)
        n_samples, n_features = features.shape
        class_counts = np.array([(labels == label).sum() for label in self.classes_])
        if np.any(class_counts == 0):
            raise ValueError("Both classes must occur in the training labels")
        self.log_class_prior_ = np.log(class_counts / n_samples)
        present_counts = np.vstack([features[labels == label].sum(axis=0) for label in self.classes_])
        self.feature_probability_ = (present_counts + self.alpha) / (class_counts[:, None] + 2 * self.alpha)
        self.log_feature_probability_ = np.log(self.feature_probability_)
        self.log_feature_absence_probability_ = np.log(1 - self.feature_probability_)
        return self

    def predict(self, features: np.ndarray) -> np.ndarray:
        log_scores = self.log_class_prior_ + (features @ self.log_feature_probability_.T + (1 - features) @ self.log_feature_absence_probability_.T)
        return self.classes_[np.argmax(log_scores, axis=1)]

def accuracy(predictions: np.ndarray, labels: np.ndarray) -> float:
    return float(np.mean(predictions == labels))
```

```python
with np.errstate(divide="ignore", over="ignore", invalid="ignore"):
    classifier = BernoulliNaiveBayes(alpha=1.0).fit(train_features, train_labels)
    train_predictions = classifier.predict(train_features)
    test_predictions = classifier.predict(test_features)

print(f"Training messages: {len(train_data)}")
print(f"Test messages:     {len(test_data)}")
print(f"Vocabulary size:   {len(vocabulary)}")
print(f"Training accuracy: {accuracy(train_predictions, train_labels):.2%}")
print(f"Testing accuracy:  {accuracy(test_predictions, test_labels):.2%}")

reference = BernoulliNB(alpha=1.0).fit(train_features, train_labels)
sklearn_predictions = reference.predict(test_features)
matches = np.array_equal(test_predictions, sklearn_predictions)
print(f"scikit-learn test accuracy: {accuracy(sklearn_predictions, test_labels):.2%}")
print(f"Predictions match scikit-learn: {matches}")
```
